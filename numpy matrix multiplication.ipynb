```python
a=np.array([0,1,2,3,4,5])
a
```

use of mask

```python
mask=(a%2==0)
a[mask]=-1
print(a)
```

aggregate function

```python
a=np.array([10,20,35,40,55])
print(np.sum(a))
print(np.mean(a))
print(np.min(a))
print(np.max(a))
```

use of all and any

```python
np.all(a<30)
```

```python
np.any(a<30)
```

where clause

```python
a=np.array([10,20,35,40,55])
b=np.where(a%2!=0,-1,1)        #syntax  where( condition,  ifTrue,  ifFalse)
b
```

sorting

```python
a=np.array([100,210,50,140,155])
b=np.sort(a)
b
```

matrix multiplication

```python
a=np.arange(1,13).reshape(3,4)
b=np.arange(2,14).reshape(4,3)
print(a,"\n\n",b)
```

methods of multiplication

```python
c=np.dot(a,b)
c
```

```python
c=a@b
c
```

```python
c=np.matmul(a,b)
c
```
